## What the figure supports

It describes reported daily mean discharge at one USGS station during a defined summer 2024 period. It can help us form questions about day-to-day and seasonal variation at that station.

## What it cannot say

It cannot establish a long-term trend, explain why discharge changed, represent all of Boulder Creek, or determine ecological, community, or water-management impacts. Values and qualification codes need review against USGS documentation before high-stakes use.

## Assumptions and uncertainty

- We assume the API returned the intended site, parameter, and period; `provenance.json` lets a reviewer verify this.
- Measurements and daily aggregation have uncertainty.
- Missing observations are excluded from the processed table, which may affect summaries.
- The selected season is illustrative and should not be generalized to other years.
