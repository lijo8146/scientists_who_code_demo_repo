# 04: Interpret limitations

## What the figure supports

It describes reported daily mean discharge at one USGS station during a defined 2024 summer period. It may help us formulate questions about day-to-day and seasonal variation.

## What it cannot say

It cannot establish a long-term trend, explain why discharge changed, represent all of Boulder Creek, or assess ecological, community, or water-management impacts. Values and qualification codes should be checked against USGS documentation before high-stakes use.

## Assumptions and uncertainty

We assume the API returned the intended site, parameter, and period; the provenance record lets a reviewer verify that. Measurements and daily aggregation have uncertainty. Missing observations are dropped from the processed table, which could affect summaries.

## Next steps

Compare multiple years with a pre-specified question, review station metadata and qualifiers, or co-design a relevant question with affected communities. If data have governance requirements, pause and use the project-specific process described in `docs/data-governance.md`.
